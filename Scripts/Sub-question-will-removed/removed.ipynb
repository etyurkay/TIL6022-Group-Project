Q5 - Full Train Consignment Trend Analysis

Importing libraries 

```python
import pandas as pd
import numpy as np
import plotly.express as px
import plotly.graph_objects as go
from scipy import stats
import statsmodels.api as sm
```

Uploading data

```python
file_path = r"C:\Users\Ecem\Desktop\TIL\1. TIL6022 - Python\Project\TIL6022-Group-Project\Data\Processed\merged_eurostat_clean_V2.csv"

df = pd.read_csv(file_path)
```

Selecting relevant columns

```python
trend_features = [
    "geo",
    "TIME_PERIOD",
    "Consignment_full_train_load_THS_T",
    "Observed_freight_total_THST"
]

q5_df = df[trend_features].copy()

eu_countries = [
    "AT","BE","BG","HR","CY","CZ","DK","EE","FI","FR","DE","EL",
    "HU","IE","IT","LV","LT","LU","MT","NL","PL","PT","RO","SK",
    "SI","ES","SE"
]

q5_df = q5_df[q5_df["geo"].isin(eu_countries)].copy()
```

Filtering Valid Countries

```python
# Her ülke için kaç yıl veri var (NaN olmayan)
years_with_data = (
    q5_df.groupby("geo")["Consignment_full_train_load_THS_T"]
    .apply(lambda s: s.notna().sum())
    .sort_values(ascending=False)
)

# Trend analizi için minimum yıl eşiği (örnek: 5 yıl)
MIN_YEARS = 17
valid_countries = years_with_data[years_with_data >= MIN_YEARS].index.tolist()
print(f"\n{len(valid_countries)} countries have ≥{MIN_YEARS} years of data.")

# Filtrelenmiş veri seti
q5_df_filtered = q5_df[q5_df["geo"].isin(valid_countries)].copy()
```

Identifying Top and Bottom 3 Countries

```python
freight_by_country = (
    q5_df_filtered.groupby("geo")["Observed_freight_total_THST"]
    .sum()
    .sort_values(ascending=False)
)

top3_countries = freight_by_country.head(3).index.tolist()
bottom3_countries = freight_by_country.tail(3).index.tolist()

print("Top 3 freight countries:", top3_countries)
print("Bottom 3 freight countries:", bottom3_countries)
```

Trending Visualization for Top and Bottom 3 Countries

```python
print("Top 3:", top3_countries)
print("Bottom 3:", bottom3_countries)

# Top 3 ülkelerin yıllara göre full-train yük trendi
fig_top = px.line(
    q5_df_filtered[q5_df_filtered["geo"].isin(top3_countries)],
    x="TIME_PERIOD",
    y="Consignment_full_train_load_THS_T",
    color="geo",
    markers=True,
    title="Trend of Full Train Consignments – Top 3 EU Freight Countries",
    labels={
        "TIME_PERIOD": "Year",
        "Consignment_full_train_load_THS_T": "Full Train Consignments (Thousand Tonnes)",
        "geo": "Country"
    }
)
fig_top.update_layout(template="plotly_white")
fig_top.show()

# Bottom 3 ülkelerin yıllara göre full-train yük trendi
fig_bottom = px.line(
    q5_df_filtered[q5_df_filtered["geo"].isin(bottom3_countries)],
    x="TIME_PERIOD",
    y="Consignment_full_train_load_THS_T",
    color="geo",
    markers=True,
    title="Trend of Full Train Consignments – Bottom 3 EU Freight Countries",
    labels={
        "TIME_PERIOD": "Year",
        "Consignment_full_train_load_THS_T": "Full Train Consignments (Thousand Tonnes)",
        "geo": "Country"
    }
)
fig_bottom.update_layout(template="plotly_white")
fig_bottom.show()
```
